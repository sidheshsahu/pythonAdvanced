```python
from pydantic import BaseModel,Field
from typing import Literal,Optional
```

```python
class Student(BaseModel):
    name: str=Field(...,description="Name of the student")
    age: int=Field(...,description="Age of the student")
    email: str=Field(...,description="Email of the student")
    gender:Literal["Male","Female","Other"]=Field(...,description="Gender of the student")
```

```python
res=Student(name="Sidhesh",age=22,email="sideshsahu@gmail.com",gender="Male")

res
```

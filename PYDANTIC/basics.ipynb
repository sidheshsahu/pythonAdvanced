```python
from pydantic import BaseModel
```

```python
class Student(BaseModel):
    name: str
    age: int
    email: str
```

```python
res=Student(name="Sidhesh",age=22,email="sideshsahu@gmail.com")
```

```python
res
```

```python
res1=Student(name="Sidhesh",age="22",email="sideshsahu@gmail.com")# Here age is string but pydantic will convert it to int automatically and pydantic tries to validate the data and convert it to the correct type if possible from its own side.
```

```python
res1
```

```python
from pydantic import StrictInt
```
